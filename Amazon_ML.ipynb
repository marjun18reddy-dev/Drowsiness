```python
import pandas as pd
import os

BASE = "/content/amazon_ml/student_resource"

train_s1 = pd.read_csv(
    f"{BASE}/dataset/train/train_source1.tsv",
    sep="\t"
)

train_s2 = pd.read_csv(
    f"{BASE}/dataset/train/train_source2.tsv",
    sep="\t"
)

train_s3 = pd.read_csv(
    f"{BASE}/dataset/train/train_source3.tsv",
    sep="\t"
)

ground_truth = pd.read_csv(
    f"{BASE}/dataset/train/train_ground_truth.tsv",
    sep="\t"
)

print("========== DATASET SHAPES ==========")
print("TRAIN SOURCE 1 :", train_s1.shape)
print("TRAIN SOURCE 2 :", train_s2.shape)
print("TRAIN SOURCE 3 :", train_s3.shape)
print("GROUND TRUTH   :", ground_truth.shape)

print("\n========== COLUMNS ==========")
print("S1:", train_s1.columns.tolist())
print("S2:", train_s2.columns.tolist())
print("S3:", train_s3.columns.tolist())
print("GT:", ground_truth.columns.tolist())

print("\n========== COUNTRIES ==========")
print("\nS1:")
print(train_s1["country"].value_counts(dropna=False))

print("\nS2:")
print(train_s2["country"].value_counts(dropna=False))

print("\nS3:")
print(train_s3["country"].value_counts(dropna=False))

print("\n========== MISSING VALUES ==========")
print("\nS1:")
print(train_s1.isna().sum())

print("\nS2:")
print(train_s2.isna().sum())

print("\nS3:")
print(train_s3.isna().sum())

print("\nGT:")
print(ground_truth.isna().sum())
```

```python
# STEP 2 — Analyze the training ground truth

import pandas as pd

# ground_truth was already loaded in Step 1

# Count how many matched entities each Source-1 entity has
def count_matches(x):
    if pd.isna(x) or str(x).strip() == "":
        return 0
    return len([v for v in str(x).split(",") if v.strip()])

ground_truth["match_count"] = ground_truth["matched_entity_ids"].apply(count_matches)

print("========== MATCH COUNT SUMMARY ==========")
print(ground_truth["match_count"].describe())

print("\n========== MATCH COUNT DISTRIBUTION ==========")
print(
    ground_truth["match_count"]
    .value_counts()
    .sort_index()
)

print("\n========== IMPORTANT COUNTS ==========")

zero_matches = (ground_truth["match_count"] == 0).sum()
one_match = (ground_truth["match_count"] == 1).sum()
multiple_matches = (ground_truth["match_count"] > 1).sum()

print("S1 entities with 0 matches :", zero_matches)
print("S1 entities with 1 match   :", one_match)
print("S1 entities with >1 match  :", multiple_matches)

print("\n========== PERCENTAGES ==========")

total = len(ground_truth)

print("0 matches  :", round(zero_matches / total * 100, 2), "%")
print("1 match    :", round(one_match / total * 100, 2), "%")
print(">1 matches :", round(multiple_matches / total * 100, 2), "%")

print("\n========== EXAMPLES ==========")

print("\nNo-match examples:")
display(
    ground_truth[
        ground_truth["match_count"] == 0
    ].head(5)
)

print("\nSingle-match examples:")
display(
    ground_truth[
        ground_truth["match_count"] == 1
    ].head(5)
)

print("\nMultiple-match examples:")
display(
    ground_truth[
        ground_truth["match_count"] > 1
    ].head(5)
)
```

```python
# STEP 3 — Validate ground-truth references

import pandas as pd

# These were already loaded in Step 1:
# train_s1
# train_s2
# train_s3
# ground_truth

print("========== BUILDING ID SETS ==========")

s1_ids = set(train_s1["entity_id"].astype(str))
s2_ids = set(train_s2["entity_id"].astype(str))
s3_ids = set(train_s3["entity_id"].astype(str))

print("S1 IDs:", len(s1_ids))
print("S2 IDs:", len(s2_ids))
print("S3 IDs:", len(s3_ids))

# Collect all matched IDs from ground truth
all_gt_matches = set()

for value in ground_truth["matched_entity_ids"].dropna():
    value = str(value).strip()

    if value:
        for entity_id in value.split(","):
            entity_id = entity_id.strip()

            if entity_id:
                all_gt_matches.add(entity_id)

print("\nTotal UNIQUE matched IDs in ground truth:", len(all_gt_matches))

# Check whether every ground-truth matched ID exists in S2 or S3
valid_s2_s3_ids = s2_ids | s3_ids

missing_ids = all_gt_matches - valid_s2_s3_ids

print("\n========== RESULT ==========")
print("Ground-truth matched IDs:", len(all_gt_matches))
print("Valid S2/S3 IDs:", len(all_gt_matches - missing_ids))
print("Missing/invalid IDs:", len(missing_ids))

if missing_ids:
    print("\nFirst 50 missing IDs:")
    print(list(missing_ids)[:50])
else:
    print("\nPASS: Every ground-truth matched ID exists in Source 2 or Source 3.")

# Also check that ground-truth Source-1 IDs exist in Source 1
gt_s1_ids = set(
    ground_truth["source1_entity_id"].astype(str)
)

missing_s1 = gt_s1_ids - s1_ids

print("\nGround-truth S1 IDs:", len(gt_s1_ids))
print("Missing S1 IDs:", len(missing_s1))

if missing_s1:
    print("First 50 missing S1 IDs:")
    print(list(missing_s1)[:50])
else:
    print("PASS: Every ground-truth Source-1 ID exists in Source 1.")
```

```python
# STEP 4 — Analyze exact matching signals
#
# IMPORTANT:
# We are NOT building the final model yet.
# We are only measuring how much of the ground truth
# can be explained by simple exact matches.

import pandas as pd
import re
import unicodedata

# ---------------------------------------------------------
# Normalization function
# ---------------------------------------------------------

def normalize_text(x):
    if pd.isna(x):
        return ""

    x = str(x).lower().strip()

    # Unicode normalization
    x = unicodedata.normalize("NFKC", x)

    # Replace punctuation with spaces
    x = re.sub(r"[^\w\s]", " ", x, flags=re.UNICODE)

    # Collapse whitespace
    x = re.sub(r"\s+", " ", x).strip()

    return x


# ---------------------------------------------------------
# Create normalized columns
# ---------------------------------------------------------

print("Creating normalized fields...")

for df in [train_s1, train_s2, train_s3]:

    df["name_norm"] = df["business_name"].apply(normalize_text)
    df["address_norm"] = df["business_address"].apply(normalize_text)
    df["country_norm"] = df["country"].fillna("").astype(str).str.lower().str.strip()


# ---------------------------------------------------------
# Build lookup dictionaries
# ---------------------------------------------------------

print("Building lookup dictionaries...")

s2_name = {}
s2_address = {}
s2_name_address = {}

for _, row in train_s2.iterrows():

    eid = row["entity_id"]

    name = row["name_norm"]
    address = row["address_norm"]

    if name:
        s2_name.setdefault(name, []).append(eid)

    if address:
        s2_address.setdefault(address, []).append(eid)

    if name and address:
        s2_name_address.setdefault(
            (name, address), []
        ).append(eid)


s3_name = {}
s3_address = {}
s3_name_address = {}

for _, row in train_s3.iterrows():

    eid = row["entity_id"]

    name = row["name_norm"]
    address = row["address_norm"]

    if name:
        s3_name.setdefault(name, []).append(eid)

    if address:
        s3_address.setdefault(address, []).append(eid)

    if name and address:
        s3_name_address.setdefault(
            (name, address), []
        ).append(eid)


# ---------------------------------------------------------
# Combine S2 + S3 lookup
# ---------------------------------------------------------

def combine_lookup(a, b):

    result = {}

    for key, values in a.items():
        result.setdefault(key, []).extend(values)

    for key, values in b.items():
        result.setdefault(key, []).extend(values)

    return result


name_lookup = combine_lookup(s2_name, s3_name)
address_lookup = combine_lookup(s2_address, s3_address)
name_address_lookup = combine_lookup(
    s2_name_address,
    s3_name_address
)


# ---------------------------------------------------------
# Ground truth lookup
# ---------------------------------------------------------

gt_lookup = {}

for _, row in ground_truth.iterrows():

    s1_id = row["source1_entity_id"]

    value = row["matched_entity_ids"]

    if pd.isna(value) or str(value).strip() == "":
        gt_lookup[s1_id] = set()
    else:
        gt_lookup[s1_id] = set(
            x.strip()
            for x in str(value).split(",")
            if x.strip()
        )


# ---------------------------------------------------------
# Evaluate exact rules
# ---------------------------------------------------------

stats = {
    "name_exact": 0,
    "address_exact": 0,
    "name_address_exact": 0,
    "name_exact_full_recall": 0,
    "address_exact_full_recall": 0,
    "name_address_exact_full_recall": 0,
}

total_gt = 0


for _, row in train_s1.iterrows():

    s1_id = row["entity_id"]

    true_matches = gt_lookup.get(s1_id, set())

    total_gt += len(true_matches)

    name_candidates = set(
        name_lookup.get(row["name_norm"], [])
    )

    address_candidates = set(
        address_lookup.get(row["address_norm"], [])
    )

    name_address_candidates = set(
        name_address_lookup.get(
            (row["name_norm"], row["address_norm"]),
            []
        )
    )

    # Number of true matches recovered
    stats["name_exact"] += len(
        true_matches & name_candidates
    )

    stats["address_exact"] += len(
        true_matches & address_candidates
    )

    stats["name_address_exact"] += len(
        true_matches & name_address_candidates
    )

    # Full recovery of this S1's ground truth
    if true_matches and true_matches.issubset(name_candidates):
        stats["name_exact_full_recall"] += 1

    if true_matches and true_matches.issubset(address_candidates):
        stats["address_exact_full_recall"] += 1

    if true_matches and true_matches.issubset(name_address_candidates):
        stats["name_address_exact_full_recall"] += 1


# ---------------------------------------------------------
# Print results
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 4 RESULTS")
print("=" * 70)

print("Total ground-truth matched pairs:", total_gt)

print("\nTRUE MATCH PAIR RECOVERY")

for rule in [
    "name_exact",
    "address_exact",
    "name_address_exact"
]:

    recovered = stats[rule]

    recall = (
        recovered / total_gt * 100
        if total_gt > 0 else 0
    )

    print(
        f"{rule:30s}: "
        f"{recovered:,} / {total_gt:,} "
        f"({recall:.2f}%)"
    )


print("\nFULL S1 RECOVERY")

for rule in [
    "name_exact_full_recall",
    "address_exact_full_recall",
    "name_address_exact_full_recall"
]:

    print(
        f"{rule:30s}: "
        f"{stats[rule]:,} / {len(train_s1):,} "
        f"({stats[rule] / len(train_s1) * 100:.2f}%)"
    )
```
